# tpd-arc2: DSL-BFS reach/selection solver (v4 + T)

Runs the frozen tpd-arc enumerator with the v4 alphabet plus the ten triangulated operations of condition T from the preregistered B5 comparison (len<=3, 1M nodes/task, R1 first-fit selection, two attempts) as a static musl binary. Source: src/bin/arc2_solve_t.rs in the reach-not-selection repository; binary sha256 3c1b0530e2fcab89106fdd730bb3a21aef0ce463c6e366209710155864c3ddd0. The arm mechanism is byte-verified against arc2_solve at empty arm, and the licensed evaluation read measured v4+T reach at 0 of 120 public evaluation tasks, so the predicted public score is 0.00; the paper reports that prediction. No internet, no models, CPU only.

```python
import shutil, subprocess, os, hashlib, json, glob

print('input tree:')
for root, dirs, files in os.walk('/kaggle/input'):
    for f in files[:20]:
        print(' ', os.path.join(root, f))

cands = glob.glob('/kaggle/input/**/arc2_solve_t', recursive=True)
assert cands, 'solver binary not found anywhere under /kaggle/input'
BIN_SRC = cands[0]
BIN = '/kaggle/working/arc2_solve_t'
chal = glob.glob('/kaggle/input/**/arc-agi_test_challenges.json', recursive=True)
assert chal, 'test challenges not found'
CHALLENGES = chal[0]
SUB = '/kaggle/working/submission.json'

shutil.copy(BIN_SRC, BIN)
os.chmod(BIN, 0o755)
print('binary sha256:', hashlib.sha256(open(BIN,'rb').read()).hexdigest())

arm = glob.glob('/kaggle/input/**/floor_T.json', recursive=True)
assert arm, 'floor_T.json not found'
r = subprocess.run([BIN, arm[0] + ':T', CHALLENGES, SUB], capture_output=True, text=True)
print(r.stderr[-2000:])
assert r.returncode == 0, r.returncode

sub = json.load(open(SUB))
print('tasks in submission:', len(sub))
print('submission sha256:', hashlib.sha256(open(SUB,'rb').read()).hexdigest())
```
